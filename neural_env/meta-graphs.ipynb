```python
# This file was used to create graphs from metacentrum output. The files had accuraccy and loss values. We try to extract these
# data and create graphs from that
import cv2
import numpy as np
from matplotlib import pyplot as plt

file_name = 'result.out'
loss_values = []
val_loss_values = []
acc_values = []
val_acc_values = []

with open(file_name, 'r') as file:
    lines1 = file.readlines()
count = 0

for line in lines1:
    if 'val_loss' not in line:
        continue

    words = line.split()

    for i, word in enumerate(words):
        # Check if the word is "loss"       
        if word == 'loss:':
            loss_value = float(words[i + 1])
            loss_values.append(loss_value)
            count += 1
        
        # Check if the word is "val_loss"
        elif word == 'val_loss:':
            val_loss_value = float(words[i + 1])
            val_loss_values.append(val_loss_value)

with open(file_name, 'r') as file:
    lines2 = file.readlines()
for line in lines2:
    if 'val_loss' not in line:
        continue
    words = line.split()
    for i, word in enumerate(words):
        # Check if the word is "loss"
        if word == 'accuracy:':
            loss_value = float(words[i + 1])
            acc_values.append(loss_value)
            count += 1
        
        # Check if the word is "val_loss"
        elif word == 'val_accuracy:':
            val_loss_value = float(words[i + 1])
            val_acc_values.append(val_loss_value)

#Plot loss
fig = plt.figure()
plt.plot(loss_values, color='teal', label='loss')
plt.plot(val_loss_values, color='orange', label='val_loss')
fig.suptitle('Loss', fontsize=20)
plt.legend(loc="upper right")
plt.show()

# Save loss plot as SVG
fig.savefig('loss_plot.svg')

# Save loss plot as PNG
fig.savefig('loss_plot.png')
```

```python
fig = plt.figure()
plt.plot(acc_values, color='teal', label='accuracy')
plt.plot(val_acc_values, color='orange', label='val_accuracy')
fig.suptitle('Accuracy', fontsize=20)
plt.legend(loc="upper left")
plt.show()

# Save loss plot as SVG
fig.savefig('accuracy_plot.svg')

# Save loss plot as PNG
fig.savefig('accuracy_plot.png')
```
