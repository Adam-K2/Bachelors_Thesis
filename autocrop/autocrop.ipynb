```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import PIL

sys.path.insert(0, Path.cwd().parent)

from autocrop import Cropper

from glob import glob

from natsort import natsorted

# Trying to sort images, before the order was random
faces_unsorted = [f for f in glob("images/*") if not f.endswith("md")]
faces = natsorted(faces_unsorted)
print(f"{len(faces)} images to test with.")

# Here we are setting to focus on face and set size of the crop image to 128x128
c = Cropper(face_percent=100,height=128,width=128)
# Setting the start number for first image and then it increments by 1
count = 559

for loc in faces:
    plt.imshow(plt.imread(loc), cmap='Greys_r')
    
    plt.imshow(c.crop(loc))
    cropped_image = c.crop(loc)
    cropped_pil_image = PIL.Image.fromarray(cropped_image)

    output_file = f"Deep4_{count}.jpg" #Setting the name of output file - Deepfakes starts with Deep and number of which deepfake algorithm was used. For original images starts with Origin
    cropped_pil_image.save(output_file)

    count += 1
```
